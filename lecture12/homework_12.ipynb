## Problem 1 - Comparing performance of robotic systems

You are considering purchasing a robotic system for manufacturing masks.
There are two possibilities, say $A$ and $B$.
They both produce the same number of masks per day, they cost the same to purchase, and the have the same power and supply costs.
However, they are not identical. They have different faulty mask rates.
Let $X_A$ and $X_B$ be the number of faulty masks you get from each system, respectively, in a given day.
For each of the possibilities below:

1) Use `scipy.stats` to make two Normal random variables $X_A$ and $X_B$ with the right mean and variance.

2) Plot the PDF of the random variables in the same figure.

3) Find a 95% central credible interval.

3) Indicate which robotic system you would buy and why (three choices $A$, $B$, and ``I cannot choose'').


+ **Case 1:** $\mathbf{E}[X_A] = 1, \mathbf{V}[X_A] = 0.1$ and $\mathbf{E}[X_B] = 1, \mathbf{V}[X_B] = 0.2$.

**Answer:**

````python
avgA1 = 1
varA1 = 0.1
avgB1 = 1
varB1 = 0.2

XA1 = st.norm(loc=avgA1,
             scale=np.sqrt(varA1))
XB1 = st.norm(loc=avgB1,
             scale=np.sqrt(varB1))
````

````python
x = np.linspace(-1, 3, 1000)
plt.figure(figsize=(8, 6))
plt.plot(x, XA1.pdf(x), label='Robot A')
plt.plot(x, XB1.pdf(x), label='Robot B')
plt.title('PDF of Faulty Masks Case 1')
plt.xlabel('Number of Faulty Masks')
plt.ylabel('Density')
plt.legend()
plt.show()
````

````python
XA1_025 = XA1.ppf(0.025)
XA1_975 = XA1.ppf(0.975)

print(f"For Case 1, the 95% credible interval for A is from {XA1_025:.3f} to {XA1_975:.3f}")
````

````python
XB1_025 = XB1.ppf(0.025)
XB1_975 = XB1.ppf(0.975)

print(f"For Case 1, the 95% credible interval for B is from {XB1_025:.3f} to {XB1_975:.3f}")
````

While both robots average 1 faulty mask per day, robot B has a greater variance. This means that robot A is likely more consistent and has a smaller likelihood of producing batches with higher numbers of defective masks. This is backed up by the confidence intervals, as the interval for A is a lot tighter than that of B. Thus, I would pick robot A.

## Problem 2 - Figuring out which household conserves less energy

In this homework problem, we are going to look at a dataset for which the Normal is not a good fit.
In particular, we are going to look at HVAC energy consumption in our high-performance building data.

````python
import requests
import os

# The url of the file we want to download
url = 'https://raw.githubusercontent.com/PurdueMechanicalEngineering/me-297-intro-to-data-science/master/data/temperature_raw.xlsx'
!curl -O $url

import numpy as np
import pandas as pd
df = pd.read_excel('temperature_raw.xlsx')
df = df.dropna(axis=0)
df.date = pd.to_datetime(df['date'], format='%Y-%m-%d')
df
````

+ Extract the `hvac` column for household `a5`:

````python
hvac_a5 = df[df['household'] == 'a5']['hvac']
````

+ Do the histogram of `hvac` for household `a5`:

````python
plt.figure(figsize=(8, 6))
plt.hist(hvac_a5)
plt.xlabel('HVAC Energy Consumption')
plt.ylabel('Frequency')
plt.title('Histogram of HVAC Data for Households a5')
plt.show()
````

+ Use the method of moments to fit a Normal distribution to the `hvac` data for household `a5`:

````python
mu = hvac_a5.mean()
std = hvac_a5.std()
HVAC_a5 = st.norm(loc=mu, scale=std)
````

+ In the same figure, show the histogram of `hvac` data for household `a5` (use `density=True` to make sure itis normalized) and the PDF of the Normal you just fitted.
Is this a good fit? Why do you think we do not get a good fit?

````python
plt.figure(figsize=(8, 6))
plt.hist(hvac_a5, density=True)

mina5 = hvac_a5.min()
maxa5 = hvac_a5.max()
xa5 = np.linspace(mina5, maxa5, 500)
plt.plot(xa5, HVAC_a5.pdf(xa5))
plt.xlabel('HVAC Energy Consumption')
plt.ylabel('Density')
plt.title('HVAC Energy Consumption for Household a5: PDF vs Histogram')
plt.show()
````

No, this is not a good fit. This is because energy consumption cannot be negative, so the data is right-skewed by default. Whereas, a normal distribution is symmetric and extends into the negative energy values, which is impossible.

+ Now I am asking you to transform the data in a way that will make them look more Normal.
Do the histogram of the logarithm of the `hvac` data for household `a5`:

````python
log_a5 = np.log(hvac_a5)

plt.figure(figsize=(8, 5))
plt.hist(log_a5)
plt.xlabel('Log(HVAC Energy Consumption)')
plt.ylabel('Frequency')
plt.title('Histogram of Log-Scaled HVAC Data for Household a5')
plt.show()
````

+ Fit a Normal to the logarithm of the `hvac` data for household `a5`:

````python
log_mu = log_a5.mean()
log_std = log_a5.std()
LOG_a5 = st.norm(loc=log_mu, scale=log_std)
````

+ In the same figure, show the histogram of the **logarithm** of the `hvac` data for household `a5` and the PDF of the Normal you just fitted. Does this look like a good fit?

````python
plt.figure(figsize=(8, 6))
plt.hist(log_a5, density=True)

minlog = log_a5.min()
maxlog = log_a5.max()
xlog = np.linspace(minlog, maxlog+1, 500)
plt.plot(xlog, LOG_a5.pdf(xlog))
plt.xlabel('Log(HVAC Energy Consumption)')
plt.ylabel('Density')
plt.title('Log-Scaled HVAC Energy Consumption for Household a5: PDF vs Histogram')
plt.show()
````

Yes, this looks like a much better fit. By applying the logarithmic scale, this fixed the issue of the PDF extending into the negative energy values. Furthermore, it smoothes out the skewed into a more bell-like shape.

+ Now do exactly the same thing as the previous bullet point for household `a3`.

````python
hvac_a3 = df[df['household'] == 'a3']['hvac']
log_a3 = np.log(hvac_a3)

log_mu3 = log_a3.mean()
log_std3 = log_a3.std()
LOG_a3 = st.norm(loc=log_mu3, scale=log_std3)

plt.figure(figsize=(8, 6))
plt.hist(log_a3, density=True)

minlog3 = log_a3.min()
maxlog3 = log_a3.max()
xlog3 = np.linspace(minlog3, maxlog3+4, 500)
plt.plot(xlog3, LOG_a3.pdf(xlog3))
plt.xlabel('Log(HVAC Energy Consumption)')
plt.ylabel('Density')
plt.title('Log-Scaled HVAC Energy Consumption for Household a3: PDF vs Histogram')
plt.show()
````

+ Which household consumes more energy, `a5` or `a3`?

````python
print(f"Mean for a5: {hvac_a5.mean():.3f}")
print(f"Mean for a3: {hvac_a3.mean():.3f}")
````

Household a3 consumes more energy.

## Problem 3 - Introducing the Log-Normal distribution

In the previous problem, we took the logarithm of `hvac` in order to obtain a better fit to the Normal.
It turns out that this is a very common practice whenever you have a positive dataset that is skewed in the way we noticed above.
As a matter of fact, there is a distribution called the [Log-Normal distribution](https://en.wikipedia.org/wiki/Log-normal_distribution) which is designed to do exactly that.
Below, I show you could have fitted a Log-Normal distribution directly on the `hvac` data.

````python
params = st.lognorm.fit(hvac_a5) # This does something similar to the method of moments
HVAC_a5_ln = st.lognorm(*params) # This is the random varibale
# Here is how you can sample from it:
HVAC_a5_ln.rvs(size=10)
````

````python
# Here is how you can evaluate its PDF:
HVAC_a5_ln.pdf(200)
````

+ In the same figure, plot the PDF of the Log-normal along with the histogram of the `hvac` data for `a5`.

````python
plt.figure(figsize=(8, 6))
plt.hist(hvac_a5, density=True)

plt.plot(xa5, HVAC_a5_ln.pdf(xa5))
plt.xlabel('HVAC Energy Consumption')
plt.ylabel('Density')
plt.title('Household a5: Raw HVAC Data vs. Log-Normal Fit')
plt.show()
````

+ Is the fit with Log-Normal good?

The log-normal is fantastic, as it preserves the original shape of the data while also fitting the data very well.

+ Recall how in [](lecture12:standard-normal-quantiles) we used the `ppf()` function of a random variable to find quantiles.
Use this function to find the 0.025-quantile of the Log-Normal we constructed above.

````python
a5ln_025 = HVAC_a5_ln.ppf(0.025)
print(f"The 0.025-quantile for Household a5 is: {a5ln_025:.3f}")
````

+ Now find the 95% central credible interval of the Log-Normal variable we constructed above.

````python
a5ln_975 = HVAC_a5_ln.ppf(0.975)
print(f"The 95% central credible interval of the Log-Normal variable for Household a5 is from {a5ln_025:.3f} to {a5ln_975:.3f}")
````

+ Repeat as many code-blocks as you need to fit a Log-Normal to the `hvac` data for household `a3` and then find the 95% central credible interval of the resulting random variable.

````python
params = st.lognorm.fit(hvac_a3) # This does something similar to the method of moments
HVAC_a3_ln = st.lognorm(*params)

plt.figure(figsize=(8, 6))
plt.hist(hvac_a3, density=True)

mina3 = hvac_a3.min()
maxa3 = hvac_a3.max()
xa3 = np.linspace(mina3, maxa3, 500)
plt.plot(xa3, HVAC_a3_ln.pdf(xa3))
plt.xlabel('HVAC Energy Consumption')
plt.ylabel('Density')
plt.title('Household a5: Raw HVAC Data vs. Log-Normal Fit')
plt.show()

a3ln_025 = HVAC_a3_ln.ppf(0.025)
a3ln_975 = HVAC_a3_ln.ppf(0.975)
print(f"The 95% central credible interval of the Log-Normal variable for Household a3 is from {a3ln_025:.3f} to {a3ln_975:.3f}")
````

+ By comparing the 95% central credible intervals constructed above, which household `a5` or `a3` consumes more energy?

Household a3 consumes mroe energy.

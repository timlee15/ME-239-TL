## Problem 1 - Calculating expectations for discrete random variables

Consider the Categorical random variable $X \sim \text{Categorical}(0.2, 0.4, 0.4)$ taking three discrete values $1, 2,$ and $3$.
Find the numerical answer for the following.
Hint: You can do it by hand or write some code. It is up to you.

+ $\mathbf{E}[X] = ?$

**Answer:** <br>
$\mathbf{E}[X] = 2.20$

````python
X = np.array([1,2,3])
p_X = np.array([0.2,0.4,0.4])

E_X = np.sum(X * p_X)
print(f"E[X] = {E_X:.2f}")
````

+ $\mathbf{E}[X^2] = ?$

**Answer:**
<br>$\mathbf{E}[X^2] = 5.40$

````python
E_X2 = np.sum(X**2 * p_X)
print(f"E[X^2] = {E_X2:.2f}")
````

+ $\mathbf{V}[X] = ?$

**Answer:**
<br>$\mathbf{V}[X] = 0.56$

````python
V_X = E_X2 - E_X**2
print(f"V[X] = {V_X:.2f}")
````

+ $\mathbf{E}[e^X] = ?$

**Answer:**
<br>$\mathbf{E}[e^X] = 11.5335$

````python
E_eX = np.sum(np.exp(X) * p_X)
print(f"E[e^X] = {E_eX:.4f}")
````

## Problem 3 - Properties of expectations

Let $X$ be a random variable with expectation $\mathbf{E}[X] = 3$ and variance $\mathbf{V}[X] = 2$.
Calculate the following expressions.

+ $\mathbf{E}[4X] = ?$

**Answer:**
<br>$\mathbf{E}[4X] = 4\cdot \mathbf{E}[X] = 4\cdot 3 = 12$

+ $\mathbf{E}[X + 2] = ?$

**Answer:**
<br>$\mathbf{E}[X+2] = \mathbf{E}[X]+2 = 3 + 2 = 5$

+ $\mathbf{E}[3X + 1] = ?$

**Answer:**
<br>$\mathbf{E}[3X+1] = 3\cdot \mathbf{E}[X] + 1 = 3\cdot 3 + 1 = 10 $

+ $\mathbf{V}[5X] = ?$

**Answer:**
<br>$\mathbf{V}[5X] = 5^2 \cdot \mathbf{V}[X] = 25 \cdot 2 = 50$

+ $\mathbf{V}[X + 3] = ?$

**Answer:**
<br>$\mathbf{V}[X+3] = \mathbf{V}[X] = 2$

+ $\mathbf{V}[2X + 1] = ?$

**Answer:**
<br>$\mathbf{V}[2X+1] = 2^2 \cdot \mathbf{V}[X] = 4 \cdot 2 = 8$

+ $\mathbf{E}[X^2] = ?$

**Answer:**
<br>$\mathbf{E}[X^2] = \mathbf{V}[X] + (\mathbf{E}[X])^2 = 2 + 3^2 = 11$
